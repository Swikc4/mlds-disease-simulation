## Shared disease comparison

These are the **working simulation inputs**. All three scenarios have four values, but the per-meeting infection probabilities are **model-derived**, not directly measured. The COVID-19 infectious duration is a modelling proxy.

| Disease | Exposed days | Infectious days | Infection chance | Meetings/day |
| :--- | ---: | ---: | ---: | ---: |
| Measles | 11 | 9 | 0.4444 | 3 |
| Influenza A | 2 | 1 | 0.4267 | 3 |
| COVID-19 (Omicron) | 3 | 4 | 0.7917 | 3 |

### How the inputs were chosen

- **Measles:** 11 latent days and 9 infectious days are CDC outbreak-simulator settings (**SOURCED model values**). Infection chance = 12 / (3 × 9) = 0.4444 (**DERIVED**).
- **Influenza A:** A study estimated 1.6 latent days and 1.0 infectious day; 1.6 is rounded to 2 for this whole-day model (**SOURCED estimates, rounded**). Infection chance = 1.28 / (3 × 1) = 0.4267 (**DERIVED**).
- **COVID-19 Omicron:** A study estimated 3.1 latent days, rounded to 3 (**SOURCED estimate, rounded**). Four infectious days is a **MODELLING PROXY**, not a universally verified duration. Infection chance = 9.5 / (3 × 4) = 0.7917 (**DERIVED**).
- **All diseases:** 3 random meetings per infectious person per day comes from our Week 2 town (**ASSUMPTION**), not disease-specific contact research.

**Important limitation:** Dividing a published reproduction number by meetings per day and infectious days is a simplified calibration. These infection-chance values are **not verified probabilities for a real-world random encounter**. The published reproduction estimates also come from different contexts. Use these numbers for a provisional simulation comparison, not as direct clinical facts. ### Expected output

Submit your simulation function and the daily results for all three working scenarios. Note that the infection probabilities are model-derived and COVID-19 infectious duration is a proxy.

```python
import random

# Working disease parameters (see the comparison table for sources and limitations).
scenarios = {
    "measles": {
        "infection_chance": 0.4444,
        "exposed_days": 11,
        "infectious_days": 9,
        "meetings_per_day": 3
    },
    "influenza": {
        "infection_chance": 0.4267,
        "exposed_days": 2,
        "infectious_days": 1,
        "meetings_per_day": 3
    },
    "covid_19": {
        "infection_chance": 0.7917,
        "exposed_days": 3,
        "infectious_days": 4,
        "meetings_per_day": 3
    }
}

# Check whether all parameters are available.
def ready_to_run(inputs):
    required = (
        "infection_chance",
        "exposed_days",
        "infectious_days",
        "meetings_per_day"
    )
    return all(inputs.get(key) is not None for key in required)

def count_states(town):
        counts = {"S":0, "E":0, "I":0, "R":0}
        for person in town:
            counts[person['state']] +=1
        return counts

# Your task: Complete this function using the Week 2 code.
def run_simulation(inputs):

    # Step 1: Get the four parameters from inputs.
    infection_chance = inputs["infection_chance"]
    exposed_days = inputs["exposed_days"]
    infectious_days = inputs["infectious_days"]
    meetings_per_day = inputs["meetings_per_day"]

    # Step 2: Set random.seed(42).
    random.seed(42)

    # Step 3: Create a fresh town of 100 people.
    town = []
    for num in range(100):
        person =  {"id": num, "state": "S", "days_exposed": 0, "days_infectious": 0}
        town.append(person)

    # Start with 99 susceptible and 1 infectious.
    town[0]["state"] = "I"
 
    # Step 4: Create lists to save the daily results.
    results = {"day": [], "S": [], "E": [], "I": [], "R": []}
    counts = count_states(town)
    results["day"].append(0)
    for state in ["S", "E", "I", "R"]:
        results[state].append(counts[state])

    # Step 5: Run the simulation for 60 days.
        # Reuse the meeting and disease progression
        # rules from Week 2.
    for day in range(1,61):
        daily_meetings = []
        newly_exposed_ids = []

        for person in town:
            if person["state"] == "I":
                for _ in range(meetings_per_day):
                    other_person = random.choice(town)
                    daily_meetings.append((person["id"], other_person["id"]))

                    #(infection passes only when the person met is "S" and the random draw is below infection_chance)
                    if other_person["state"] == "S":
                        if random.random() < infection_chance:
                            newly_exposed_ids.append(other_person["id"])
        
        #5. After all meetings, set the newly exposed people to "E" with days_exposed = 0
        for pid in newly_exposed_ids:
            town[pid]["state"] = "E"
            town[pid]["days_exposed"] = 0
            town[pid]["days_infectious"] = 0

        # 6. Advance the timers: exposed people become infectious after 5 days, infectious people recover after 10 days
        for person in town:
            if person["state"] == "E":
                person["days_exposed"] += 1
                if person["days_exposed"] >= exposed_days:
                    person["state"] = "I"
                    person["days_exposed"] = 0
                    person["days_infectious"] = 0
            elif person["state"] == "I":
                person["days_infectious"] += 1
                if person["days_infectious"] >= infectious_days:
                    person["state"] = "R"
                    person["days_exposed"] = 0
                    person["days_infectious"] = 0

        # Step 6: Count each health state after every day.
        counts = count_states(town)
        results["day"].append(day)
        for state in ["S", "E", "I", "R"]:
            results[state].append(counts[state])

    # Step 7: Return the results.

    return results

test = run_simulation({"infection_chance": 0.15, "exposed_days": 5,
                       "infectious_days": 10, "meetings_per_day": 3})
print(test["S"])
print(test["E"])
print(test["I"])
print(test["R"])
```
