```python
# Create emply lists: sir_days, sir_S, sir_I, and sir_R
sir_days = []
sir_S = []
sir_I = []
sir_R = []

# Set Values:
S = 99
I = 1
R = 0
N = 100
beta = 0.3
gamma = 0.1

#Save Day 0 values to the list first:
sir_days.append(0)
sir_S.append(S)
sir_I.append(I)
sir_R.append(R)

# Create loop to repeat Day 1 calculations 30 times
for day in range(1,31):

# Ensure each calculation of new infections and new recoveries use the current S and I before changing them
    new_infections = beta * S * I / N
    new_recoveries = gamma * I

# Update S, I, and R using the same steps shown for Day 1
    S = S - new_infections
    I = I + new_infections - new_recoveries
    R = R + new_recoveries
# Check the total = 100
    if round(S + I + R, 6) != 100:
        print("S + I + R does not equal 100!")

# Save the day number and the new S, I, and R values to their list using append()
    sir_days.append(day)
    sir_S.append(S)
    sir_I.append(I)
    sir_R.append(R)

print(sir_days)
print(sir_S)
print(sir_I)
print(sir_R)
```
