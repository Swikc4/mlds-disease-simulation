## 1. What is SIR?

SIR is a simple mathematical model for how an infectious disease spreads. It divides a population into three groups:

- **S — Susceptible:** People who could become infected.
- **I — Infected:** People who are currently infectious.
- **R — Recovered/removed:** People who are no longer infectious and cannot be infected again *in this model*.

People move **S → I → R**. For now, we assume no births, deaths, or repeat infections.

**Example:** Out of 100 people, 99 are susceptible, 1 is infected, and 0 have recovered. That means **99 + 1 + 0 = 100**.

**Think about it:** If one susceptible person gets infected, which group increases and which decreases?

## 3. How do people move between groups?

Two model settings help us calculate daily changes:

- **beta (β)** controls how quickly infections spread in our simplified model.
- **gamma (γ)** controls how quickly infected people recover.

We use **beta = 0.30** and **gamma = 0.10** as *teaching assumptions*, not real rates for any particular disease.

One-day formulas:

`new_infections = beta * S * I / population`

`new_recoveries = gamma * I`

For S=99 and I=1, expected new infections = **0.297** and expected recoveries = **0.1**. Decimals are expected in a mathematical model; they do not represent fractional real people.

After calculating both from the same day's values: subtract infections from S, add infections and subtract recoveries from I, and add recoveries to R. Try one day on paper first.

```python
# WORKED EXAMPLE — one day
population = 100
S, I, R = 99.0, 1.0, 0.0
beta, gamma = 0.30, 0.10
new_infections = beta * S * I / population
new_recoveries = gamma * I
S = S - new_infections
I = I + new_infections - new_recoveries
R = R + new_recoveries
print('Day 1:', round(S,3), round(I,3), round(R,3))
print('Total:', round(S+I+R,3))
```
