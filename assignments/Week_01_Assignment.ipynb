## Brief history

Scientists wanted to understand why epidemics grow quickly, reach a peak, and then slow down. They began using mathematics to study how infections spread through a population.

During a plague outbreak in Bombay (now Mumbai), India, in 1905–1906, weekly death records showed a rise and then a fall in deaths. In 1927, scientist William O. Kermack and physician Anderson G. McKendrick used these records to compare a real outbreak with their mathematical calculations. Their model grouped people by whether they could catch an infection, were infectious, or had left the infectious group, and calculated how those groups could change over time.

Their work helped explain why an epidemic can decline even when some people have never been infected: fewer susceptible people remain for the disease to spread to. This became a foundation for the SIR model.

As scientists studied diseases with a delay between infection and becoming contagious, they added an **Exposed (E)** stage to SIR so the SEIR model could represent that waiting period.

**Sources:** [Kermack and McKendrick (1927), *A Contribution to the Mathematical Theory of Epidemics*, page 714](https://doi.org/10.1098/rspa.1927.0118) (Bombay plague records and SIR foundation); [Burke (2024), *Origins of the problematic E in SEIR epidemic models*](https://pmc.ncbi.nlm.nih.gov/articles/PMC11024649/) (history of the exposed stage in SEIR).

## Understanding SIR and SEIR Models

Both models show how people move between groups as an infection spreads. SIR has three groups, while SEIR includes an additional exposed stage.

### SIR model

$$\boxed{\text{Susceptible (S)}}\;\longrightarrow\;\boxed{\text{Infectious (I)}}\;\longrightarrow\;\boxed{\text{Recovered (R)}}$$

### SEIR model

$$\boxed{\text{Susceptible (S)}}\;\longrightarrow\;\boxed{\text{Exposed (E)}}\;\longrightarrow\;\boxed{\text{Infectious (I)}}\;\longrightarrow\;\boxed{\text{Recovered (R)}}$$

### What each group means

**Susceptible (S):** People who have not been infected but could catch the disease.

**Exposed (E):** People who are infected but not yet infectious in our simple SEIR model.

**Infectious (I):** People who can currently spread the disease.

**Recovered or removed (R):** People who are no longer infectious.

## Starting scenario

$N$ is the total population.

Population equations:

$$S+I+R=N$$

$$S+E+I+R=N$$

Our models assume a closed population without reinfection. They are mathematical teaching examples, not forecasts of actual outbreaks.
