## Brief history

In the early 1900s, scientists wanted to understand why infectious disease outbreaks grew quickly and then declined. **William O. Kermack and A. G. McKendrick** studied this question using mathematics. In their 1927 paper, *A Contribution to the Mathematical Theory of Epidemics*, they compared a mathematical epidemic curve with records of **plague deaths in Bombay (now Mumbai), India, from 1905 to 1906**.

Their approach divided a population into people who could still become infected, people who were infectious, and people who had recovered or otherwise left the infectious group. The mathematical model helped show how an epidemic could slow down as the number of susceptible people decreased, even when not everyone had been infected.

This research became a foundation for the **SIR model**. Later models such as **SEIR** added an exposed stage to represent a delay before an infected person becomes infectious. The Bombay example helped illustrate their mathematical theory, although actual plague transmission also involved rats and fleas, which a basic SIR model does not describe.

**Source:** [Kermack and McKendrick (1927)](https://doi.org/10.1098/rspa.1927.0118)

## SIR model

$$\boxed{\text{Susceptible (S)}}\;\longrightarrow\;\boxed{\text{Infectious (I)}}\;\longrightarrow\;\boxed{\text{Recovered (R)}}$$

## SEIR model

$$\boxed{\text{Susceptible (S)}}\;\longrightarrow\;\boxed{\text{Exposed (E)}}\;\longrightarrow\;\boxed{\text{Infectious (I)}}\;\longrightarrow\;\boxed{\text{Recovered (R)}}$$

### What the letters mean

**S: Susceptible.** People who have not caught the disease but could become infected.

**E: Exposed.** People who are infected but cannot spread it yet in our simple SEIR model.

**I: Infectious.** People who can currently spread the disease.

**R: Recovered or removed.** People who are no longer infectious.

SIR uses S, I, and R. SEIR adds E between S and I.
